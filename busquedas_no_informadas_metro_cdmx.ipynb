```python
from collections.abc import Iterable


class Problem[S, A]:
    def __init__(self, initial_state, goal_state) -> None:
        self.initial_state = initial_state
        self.goal_state = goal_state

    def get_actions(self, state: S) -> Iterable[S]:
        raise NotImplementedError

    def apply_action(self, state: S, action: A) -> S:
        raise NotImplementedError

    def goal_test(self, state: S) -> bool:
        return self.goal_state == state

    def action_cost(self, state1: S, action: A, state2: S) -> float:
        return 1
```

```python
class GraphProblem[S, A](Problem[S, A]):
    def __init__(
        self, initial_state: S, goal_state: S, graph: dict[S, dict[A, float]]
    ) -> None:
        super().__init__(initial_state, goal_state)
        self.graph: dict[S, dict[A, float]] = graph

    def get_actions(self, state: S) -> list[S]:
        return [key for key in self.graph[state]]

    def apply_action(self, state: S, action: A) -> S:
        return action

    def action_cost(self, state1, action, state2) -> float:
        return self.graph[state1][state2]
```

```python
class Node[S, A]:
    def __init__(
        self,
        state: S,
        parent: "Node[S, A] | None" = None,
        action: A | None = None,
        path_cost: float = 0,
    ) -> None:
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost

    def path(self) -> list[S]:
        path_list: list[Node[S]] = []

        node: Node = self
        while node:
            path_list.append(node.state)
            node: Node = node.parent

        path_list.reverse()

        return path_list

    def child_node(self, problem: GraphProblem, action: A) -> "Node[S]":
        next_state = problem.apply_action(self.state, action)
        step_cost = problem.action_cost(self.state, action, next_state)

        return Node(next_state, self, action, self.path_cost + step_cost)

    def expand(self, problem: GraphProblem) -> "list[Node[S]]":
        return [
            self.child_node(problem, action)
            for action in problem.get_actions(self.state)
        ]
```

```python
def depth_first_search[S](problem: Problem) -> Node[S]:
    start_node = Node(problem.initial_state)

    if problem.goal_test(start_node.state):
        return start_node

    frontier: list[Node[S]] = [start_node]
    explored: set[S] = set()

    while frontier:
        node: Node[S] = frontier.pop()
        explored.add(node.state)

        if problem.goal_test(node.state):
            return node

        for child in node.expand(problem):
            if child.state not in explored:
                frontier.append(child)

    return None
```

```python
from collections import deque


def breadth_first_search[S](problem: Problem) -> Node[S]:
    start_node = Node(problem.initial_state)

    if problem.goal_test(start_node.state):
        return start_node

    frontier: deque[Node[S]] = deque([start_node])
    explored: set[S] = set()

    while frontier:
        node: Node[S] = frontier.popleft()
        explored.add(node.state)

        if problem.goal_test(node.state):
            return node

        for child in node.expand(problem):
            if child.state not in explored:
                frontier.append(child)

    return None
```
